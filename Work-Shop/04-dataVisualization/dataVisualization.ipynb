```python
# Create index of 100 students and save it in a list
index_student = [idx+1 for idx in range(len(gender))]
#print(index_student) #นักเรียน 100 คน

#ดึงค่าคะแนนของนักเรียนทั้ง 3 วิชามาจากDataFrameของPandas
math_score = csv_data['math score']
reading_score = csv_data['reading score']
writing_score = csv_data['writing score']
```

```python
#เรียงค่าคะแนนวิชาคณิตศาตร์ และวิชาการอ่านจากน้อยไปหามาก
#วิชาคณิตศาสตร์
#Write code below please.
lowest_10_math_scores = math_score.sort_values().head(10).reset_index(drop=True)

#วิชาการอ่าน
#Write code below please.
lowest_10_reading_scores = reading_score.sort_values().head(10).reset_index(drop=True)
```

```python
plt.figure(figsize=(10,6))
# Write code below please. #
plt.plot(range(1, 11), lowest_10_math_scores, label='Math', color='red')
plt.plot(range(1, 11), lowest_10_reading_scores, label='Reading', color='blue')

############################
plt.title('Line chart comparison on Math and Reading score of 10 students')
plt.ylabel('Score')
plt.xlabel('Individual student')
plt.legend()
plt.grid(True)
plt.show()
```

# TO DO 5: หาค่ามัธยฐานและค่าเฉลี่ยของแต่ละวิชา และนำไปแสดงผลบนกราฟแท่งแบบ Multiple plot

<p> ภาพตัวอย่าง</p>

image.png

```python
#Write code below please.
mean_math = csv_data['math score'].mean()
mean_reading = csv_data['reading score'].mean()
mean_writing = csv_data['writing score'].mean()

median_math = csv_data['math score'].median()
median_reading = csv_data['reading score'].median()
median_writing = csv_data['writing score'].median()

print(f"Mean Math Score: {mean_math:.2f}")
print(f"Median Math Score: {median_math:.2f}")
print(f"Mean Reading Score: {mean_reading:.2f}")
print(f"Median Reading Score: {median_reading:.2f}")
print(f"Mean Writing Score: {mean_writing:.2f}")
print(f"Median Writing Score: {median_writing:.2f}")
```

```python
#Plot multiple bar graph
#Write code below please.

labels = ['Math', 'Reading', 'Writing']
means = [mean_math, mean_reading, mean_writing]
medians = [median_math, median_reading, median_writing]

x = list(range(len(labels)))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar([pos - width/2 for pos in x], means, width, label='Mean', color='#D6A2E8')
ax.bar([pos + width/2 for pos in x], medians, width, label='Median', color='#56E6C1')

ax.set_ylabel('Scores')
ax.set_title('Mean and Median Scores by Subject')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()

plt.tight_layout()
plt.show()
```
